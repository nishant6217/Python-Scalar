```python
# print prime or non prime from range

n = int(input())
flag = 0

if n <= 1:
    print("non prime")
else:
    for i in range(2, n):
        if n % i == 0:
            flag = 1
            break

    if flag == 0:
        print("prime")
    else:
        print("non prime")
```

```python
# print each digits from a number

x = int(input())

while x > 0:
    digit = x % 10
    x //= 10
    print(digit, end=" ")
```

```python
# print sum of digits of no
x = int(input())
c=0
while x > 0:
    digit = x % 10
    x //= 10
    c+=digit
print(c)
```

```python
# add a single digit to last

x = int(input())
y= int(input())

print(x*10+y)
```

```python
# print reverse of int

n = int(input())

plusMinus = 1
if n<0:
  plusMinus =0
  n = n*(-1)

rev=0
while n>0:
  endDig = n%10
  rev= rev*10 + endDig
  n//=10

if(plusMinus==0):
  rev = rev*(-1)
print(rev)
```

```python
# print reverse of int and test for multiple test casse

f = int(input())

for _ in range(f):
  n = int(input())

  plusMinus = 1
  if n<0:
    plusMinus =0
    n = n*(-1)

  rev=0
  while n>0:
    endDig = n%10
    rev= rev*10 + endDig
    n//=10

  if(plusMinus==0):
    rev = rev*(-1)
  print(rev)
```
